## Background and Motivation.

We think that the quality of our air here in the valley is often a topic discussion. However, the primary motivation came from one of our assignments. There we attempted to use the AQI to give us a moving average in order to be able to issue a warning to sensitive groups when an upwards trend was most likely occuring. We wanted to expand on that, so in this project we want to see if we can predict the number of days the AQI is at 101 or higher, which are days where the air is unhealthy for sensitive groups or higher. We do not have a backround nor have we done any research in the area, but it seemed like an interesting problem. Additionally, we want to find out through mining of API's like Facebook, how the population in general feels about a few of the ways we have been tring to clean up our air and minimize the number of days the AQI is over 101.  

## Project Objectives.

We would like to know how population growth in the valley will be affecting the number of days the AQI will reach over 101, therefore creating a forcast for future years based primarely on population growth. Population growth seems like a good measure for the increasing days of unhealthy air since it drives the amount of energy used, for example from heating and air conditioning of homes and cars on the roads in the valley. If we can create a good predictive model, then it can be used as supporting evidance for a need to make a valley wide plan for future energy consuption that could limit the amount of polutants like PM2.5 which accumulate due to our natural inversion conditions. Additionally, we would like inform our discussion by gaging attitudes for some of the measures which have been implemented or could be implemented, such as limiting use of wood buring fireplaces, addition of more public transportation, and other measures during mandatory action days.
